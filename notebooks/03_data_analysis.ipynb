#  تحليل بنية البيانات في MongoDB

هذا الدفتر (Notebook) يتصل بقاعدة البيانات الخاصة بمشروعنا `midterm_data_pipeline` ليقوم بسحب أمثلة حية للسجلات التي تمت معالجتها.
سنستعرض 3 أنواع من السجلات:
1. سجل سليم (Valid)
2. سجل مُصحح (Corrected)
3. سجل مرفوض (Quarantine)

```python
from pymongo import MongoClient
import pprint

# الاتصال بقاعدة البيانات
client = MongoClient('mongodb://localhost:27017')
db = client['midterm_data_pipeline']

val_col = db['orders_validated']
quar_col = db['orders_quarantine']

print("✅ Connected to MongoDB successfully!")
```

## 1. سجل سليم تماماً (Valid Record)

```python
# جلب سجل سليم تماماً
valid_doc = val_col.find_one({"quality_status": "VALID"})
print("\n=== VALID RECORD ===")
pprint.pprint(valid_doc)
```

## 2. سجل فيه أخطاء وتمت معالجته (Corrected Record)

```python
# جلب سجل تم تصحيحه
corrected_doc = val_col.find_one({"quality_status": "CORRECTED"})
print("\n=== CORRECTED RECORD ===")
pprint.pprint(corrected_doc)
```

## 3. سجل فاسد وتم رفضه (Quarantine Record)

```python
# جلب سجل تم عزله
quarantine_doc = quar_col.find_one({})
print("\n=== QUARANTINE RECORD ===")
pprint.pprint(quarantine_doc)
```
